```python
import numpy as np

def Lambda(m, num_points=5000):
    """
    Compute the Lebesgue constant for (m+1) equally spaced nodes
    on [-1,1].

    Parameters
    m : int
    
    num_points : int
        Number of evaluation points for approximating the infinity norm.
        Returns
    
    float
        Approximation of the Lebesgue constant.
    """
    if m < 2:
        raise ValueError 

    # Equally spaced nodes including boundaries
    x_nodes = np.linspace(-1, 1, m + 1)

    x_eval = np.linspace(-1, 1, num_points)

    # Lagrange basis values
    L = np.ones((m + 1, num_points))

    for j in range(m + 1):
        for k in range(m + 1):
            if k != j:
                L[j] *= (x_eval - x_nodes[k]) / (x_nodes[j] - x_nodes[k])

    # Lebesgue function
    lebesgue_function = np.sum(np.abs(L), axis=0)

    # Infinity norm approximation
    return np.max(lebesgue_function)

for m in [2, 5, 10]:
    print(f"m = {m},  Lambda(m) ≈ {Lambda(m):.6f}")
```

```python
import matplotlib.pyplot as plt

# Compute Lebesgue constants
m_values = np.arange(2, 31)
Lambda_values = np.array([Lambda(m) for m in m_values])

# Lebesgue constants
m_values = np.arange(2, 31)
Lambda_values = np.array([Lambda(m) for m in m_values])

g = 2**m_values / (m_values * np.log(m_values))

# Scale g to match magnitude roughly (visual comparison only)
scale = Lambda_values[-1] / g[-1]
g_scaled = scale * g

plt.figure()
plt.plot(m_values, Lambda_values)
plt.plot(m_values, g_scaled)
plt.yscale("log")
plt.xlabel("m")
plt.ylabel("Lebesgue constant (log scale)")
plt.title("Growth of Lebesgue Constant (Equally Spaced Nodes)")
plt.show()
```

The Lebesgue constant acts like a condition number for interpolation, measuring how strongly data errors are amplified. Since the expression grows exponentially for equally spaced nodes, the interpolation problem may be unstable for large 𝑚. so small perturbations may make convergence not work even though it is well-posed. 

```python
def LambdaC(m, num_points=5000):
    """
    Compute the Lebesgue constant for Chebyshev nodes
    x_i = cos(i*pi/m),  i = 0,...,m.
    """
    if m < 2:
        raise ValueError("m must be >= 2")

    # Chebyshev–Lobatto nodes
    i = np.arange(m + 1)
    x_nodes = np.cos(i * np.pi / m)

    # Evaluation grid
    x_eval = np.linspace(-1, 1, num_points)

    # Lagrange basis
    L = np.ones((m + 1, num_points))
    for j in range(m + 1):
        for k in range(m + 1):
            if k != j:
                L[j] *= (x_eval - x_nodes[k]) / (x_nodes[j] - x_nodes[k])

    lebesgue_function = np.sum(np.abs(L), axis=0)
    return np.max(lebesgue_function)

m_values = np.linspace(2, 200, 20, dtype=int)
LambdaC_values = np.array([LambdaC(m) for m in m_values])

theory = (2/np.pi) * np.log(m_values)

# Shift by additive constant for visual comparison
shift = LambdaC_values[0] - theory[0]
theory_shifted = theory + shift

plt.figure()
plt.plot(m_values, LambdaC_values)
plt.plot(m_values, theory_shifted)
plt.xlabel("m")
plt.ylabel("Lebesgue constant")
plt.title("Lebesgue Constant for Chebyshev Nodes")
plt.show()
```

The Lebesgue constant for Chebyshev nodes grows only logarithmically, kinf of like ln(m). The slower growth means that the interpolation on Chebyshev grids stays stable even for larger m. Therefore, Chebyshev nodes lead to a well-conditioned, convergent interpolation process (not as much like equidstant nodes). 

```python
import numpy as np

def recursive_f(f, nodes):
    
    nodes = np.array(nodes)
    
    # Base case: single node
    if len(nodes) == 1:
        return f(nodes[0])
    
    # Recursive step
    return (recursive_f(f, nodes[1:]) - 
            recursive_f(f, nodes[:-1])) / (nodes[-1] - nodes[0])

result = recursive_f(np.sin, [0, 1, 2, 3])
print("f[0,1,2,3] =", result)

```

```python
import numpy as np

def qk(f, nodes, k, x):
    """
    Compute the Newton interpolation polynomial q_k(x).

    Parameters
    f : callable
        Function to interpolate
    nodes : array-like
        Nodes [x0, ..., xm]
    k : int
        Degree (use nodes x0,...,xk)
    x : array-like or float
        Evaluation points
    Returns q_k(x)
    """
    nodes = np.array(nodes)
    x = np.array(x)
    
    result = recursive_f(f, [nodes[0]]) * np.ones_like(x)
    
    # Add higher-order terms
    for j in range(1, k + 1):
        coeff = recursive_f(f, nodes[:j+1])
        
        product = np.ones_like(x)
        for i in range(j):
            product *= (x - nodes[i])
        
        result += coeff * product
        
    return result

import matplotlib.pyplot as plt

# Function and nodes
f = np.sqrt
nodes = np.array([0, 1, 2, 3, 4])

# Evaluation grid
x_plot = np.linspace(0, 5, 500)

plt.figure()

# Plot q_k for k=0,...,4
for k in range(5):
    plt.plot(x_plot, qk(f, nodes, k, x_plot), label=f"q_{k}(x)")

# Plot true function
plt.plot(x_plot, f(x_plot), linestyle="--", label="sqrt(x)")

plt.legend()
plt.xlabel("x")
plt.ylabel("y")
plt.title("Recursive Newton Interpolation for sqrt(x)")
plt.show()
```

```python
# Midpoint rule
def midpoint(f, a, b):
    return (b - a) * f((a + b) / 2)

# Trapezoidal rule
def trapezoidal(f, a, b):
    return (b - a) * (f(a) + f(b)) / 2

# Simpson's rule
def simpson(f, a, b):
    return (b - a) / 6 * (f(a) + 4*f((a + b)/2) + f(b))

f = np.tanh
a, b = 0, 1

# Exact value
I_exact = np.log(np.cosh(1))

# Approximations
methods = {
    "Midpoint": midpoint,
    "Trapezoidal": trapezoidal,
    "Simpson": simpson
}

results = []

for name, method in methods.items():
    I_num = method(f, a, b)
    error = abs(I_num - I_exact)
    results.append((name, I_num, error))

# Print table (3 significant digits)
print(f"{'Method':<12} {'Value':<12} {'Error'}")
print("-"*35)
for name, val, err in results:
    print(f"{name:<12} {val:.3g} {err:.3g}")
```

Simpson’s rule is more accurate than the midpoint and trapezoidal rules because it is exact for polynomials up to degree three and then has higher-order error. The midpoint rule deos reasonably well, while the trapezoidal rule shows the largest error. shows the increasing order of accuracy of the three methods.

```python
import numpy as np
import matplotlib.pyplot as plt

# Function
f = lambda x: 1/(1 + x**2)

# h values
k_values = np.arange(0, 11)
h_values = 2.0**(-k_values)

# Store errors
error_mid = []
error_trap = []
error_simp = []

for h in h_values:
    exact = np.arctan(h)
    
    I_mid = midpoint(f, 0, h)
    I_trap = trapezoidal(f, 0, h)
    I_simp = simpson(f, 0, h)
    
    error_mid.append(abs(I_mid - exact))
    error_trap.append(abs(I_trap - exact))
    error_simp.append(abs(I_simp - exact))

error_mid = np.array(error_mid)
error_trap = np.array(error_trap)
error_simp = np.array(error_simp)

plt.figure()

plt.loglog(h_values, error_mid, label="Midpoint")
plt.loglog(h_values, error_trap, label="Trapezoidal")
plt.loglog(h_values, error_simp, label="Simpson")

# Reference slopes
plt.loglog(h_values, h_values**2, linestyle='--', label=r"$h^2$")
plt.loglog(h_values, h_values**4, linestyle='--', label=r"$h^4$")

plt.xlabel("h")
plt.ylabel("Absolute error")
plt.title("Error scaling of quadrature rules")
plt.legend()
plt.show()
```

From the log-log plot, the midpoint and trapezoidal rules scale like O(h^2), while Simpson’s rule scales like O(h^4). This confirms the theoretical error orders derived in the lecture: midpoint and trapezoidal rules are second-order accurate, and Simpson’s rule is fourth-order accurate.
